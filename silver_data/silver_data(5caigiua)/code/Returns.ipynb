```python
import pandas as pd

# 1. Đọc file dữ liệu đổi trả gốc
df_return = pd.read_csv('returns.csv')
print(f'Số dòng ban đầu (return): {len(df_return)}')

# 2. Xóa các dòng bị trùng lặp hoàn toàn
df_return = df_return.drop_duplicates()

# 3. Làm sạch chuỗi văn bản (strip khoảng trắng thừa)
text_cols_return = df_return.select_dtypes(include=['object']).columns
for col in text_cols_return:
    df_return[col] = df_return[col].astype(str).str.strip()

# 4. Xử lý giá trị khuyết thiếu (Missing values)
if 'return_id' in df_return.columns:
    df_return = df_return.dropna(subset=['return_id'])

# 5. Chuyển đổi kiểu dữ liệu cho đúng chuẩn
if 'return_quantity' in df_return.columns:
    df_return['return_quantity'] = pd.to_numeric(
        df_return['return_quantity'], errors='coerce'
    ).fillna(0).astype(int)

if 'refund_amount' in df_return.columns:
    df_return['refund_amount'] = pd.to_numeric(
        df_return['refund_amount'].astype(str).str.replace(',', ''), errors='coerce'
    )

if 'return_date' in df_return.columns:
    df_return['return_date'] = pd.to_datetime(df_return['return_date'], errors='coerce')

# 6. Chuẩn hóa tên cột về dạng snake_case
df_return.columns = (
    df_return.columns.str.strip().str.lower().str.replace(' ', '_')
)

# 7. Xuất ra file return.csv đã làm sạch chuẩn
df_return.to_csv('return.csv', index=False, encoding='utf-8-sig')

print(f'✅ Đã làm sạch và tạo xong file: return.csv (còn {len(df_return)} dòng)')
print(df_return.head())
```
