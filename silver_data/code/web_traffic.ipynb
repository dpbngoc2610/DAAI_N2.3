```python
# ============================================================
# 1. IMPORT THƯ VIỆN
# ============================================================

import pandas as pd
import numpy as np


# ============================================================
# 2. ĐỌC DỮ LIỆU GỐC
# ============================================================

file_path = "web_traffic.csv"

df = pd.read_csv(file_path)

print("Đọc dữ liệu thành công!")
print("Kích thước dữ liệu:", df.shape)

display(df.head(5))


# ============================================================
# 3. KIỂM TRA DỮ LIỆU GỐC
# ============================================================

print("========== THÔNG TIN DỮ LIỆU GỐC ==========")

print("\n1. Kích thước dữ liệu:")
print(df.shape)

print("\n2. Thông tin các cột:")
df.info()

print("\n3. Số lượng giá trị thiếu:")
print(df.isnull().sum())

print("\n4. Số dòng bị trùng:")
print(df.duplicated().sum())

print("\n5. Số ngày bị trùng:")
print(df["date"].duplicated().sum())

print("\n6. Các nguồn truy cập:")
print(df["traffic_source"].value_counts())


# ============================================================
# 4. SAO CHÉP DỮ LIỆU ĐỂ TIẾN HÀNH LÀM SẠCH
# ============================================================

df_clean = df.copy()


# ============================================================
# 5. CHUẨN HÓA TÊN CỘT
# ============================================================

df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.lower()
)

print("Tên cột sau khi chuẩn hóa:")
print(df_clean.columns.tolist())


# ============================================================
# 6. CHUYỂN ĐỔI KIỂU DỮ LIỆU NGÀY
# ============================================================

df_clean["date"] = pd.to_datetime(
    df_clean["date"],
    errors="coerce"
)


# ============================================================
# 7. CHUYỂN ĐỔI CÁC CỘT SỐ
# ============================================================

numeric_columns = [
    "sessions",
    "unique_visitors",
    "page_views",
    "bounce_rate",
    "avg_session_duration_sec"
]

for column in numeric_columns:
    df_clean[column] = pd.to_numeric(
        df_clean[column],
        errors="coerce"
    )


# ============================================================
# 8. CHUẨN HÓA CỘT TRAFFIC_SOURCE
# ============================================================

df_clean["traffic_source"] = (
    df_clean["traffic_source"]
    .astype("string")
    .str.strip()
    .str.lower()
)


# ============================================================
# 9. KIỂM TRA GIÁ TRỊ THIẾU TRƯỚC KHI XỬ LÝ
# ============================================================

print("\n========== GIÁ TRỊ THIẾU TRƯỚC KHI XỬ LÝ ==========")
print(df_clean.isnull().sum())


# ============================================================
# 10. XÓA CÁC DÒNG THIẾU DỮ LIỆU QUAN TRỌNG
# ============================================================

required_columns = [
    "date",
    "traffic_source",
    "sessions",
    "unique_visitors",
    "page_views",
    "bounce_rate",
    "avg_session_duration_sec"
]

df_clean = df_clean.dropna(
    subset=required_columns
)


# ============================================================
# 11. XÓA CÁC DÒNG BỊ TRÙNG LẶP HOÀN TOÀN
# ============================================================

df_clean = df_clean.drop_duplicates()


# ============================================================
# 12. XỬ LÝ TRÙNG NGÀY
# ============================================================

# Trong thiết kế DAILY_WEB_TRAFFIC,
# date được sử dụng làm khóa chính.
# Vì vậy, mỗi ngày chỉ được xuất hiện một lần.

df_clean = df_clean.drop_duplicates(
    subset=["date"],
    keep="first"
)


# ============================================================
# 13. LOẠI BỎ DỮ LIỆU KHÔNG HỢP LỆ
# ============================================================

# Các chỉ số đếm không được âm
df_clean = df_clean[
    (df_clean["sessions"] >= 0) &
    (df_clean["unique_visitors"] >= 0) &
    (df_clean["page_views"] >= 0)
]


# Bounce rate phải nằm trong khoảng từ 0 đến 1
df_clean = df_clean[
    (df_clean["bounce_rate"] >= 0) &
    (df_clean["bounce_rate"] <= 1)
]


# Thời gian truy cập trung bình không được âm
df_clean = df_clean[
    df_clean["avg_session_duration_sec"] >= 0
]


# ============================================================
# 14. SẮP XẾP DỮ LIỆU THEO NGÀY
# ============================================================

df_clean = df_clean.sort_values(
    by="date"
).reset_index(drop=True)


# ============================================================
# 15. ĐỊNH DẠNG LẠI CỘT DATE
# ============================================================

df_clean["date"] = df_clean["date"].dt.date


# ============================================================
# 16. KIỂM TRA DỮ LIỆU SAU KHI LÀM SẠCH
# ============================================================

print("\n========== KẾT QUẢ SAU KHI LÀM SẠCH ==========")

print("\n1. Kích thước dữ liệu:")
print(df_clean.shape)

print("\n2. Số dòng dữ liệu:")
print(len(df_clean))

print("\n3. Số giá trị thiếu:")
print(df_clean.isnull().sum())

print("\n4. Số dòng trùng lặp:")
print(df_clean.duplicated().sum())

print("\n5. Số ngày duy nhất:")
print(df_clean["date"].nunique())

print("\n6. Ngày bắt đầu:")
print(df_clean["date"].min())

print("\n7. Ngày kết thúc:")
print(df_clean["date"].max())

print("\n8. Các nguồn truy cập:")
print(df_clean["traffic_source"].unique())

print("\n9. Kiểu dữ liệu:")
print(df_clean.dtypes)

print("\n10. Năm dòng dữ liệu sau khi làm sạch:")
display(df_clean.head())


# ============================================================
# 17. KIỂM TRA TÍNH HỢP LỆ CỦA DỮ LIỆU
# ============================================================

assert df_clean["date"].is_unique
assert df_clean["date"].notna().all()

assert df_clean["sessions"].ge(0).all()
assert df_clean["unique_visitors"].ge(0).all()
assert df_clean["page_views"].ge(0).all()

assert df_clean["bounce_rate"].between(0, 1).all()

assert df_clean["avg_session_duration_sec"].ge(0).all()

print("\nDữ liệu đã vượt qua các kiểm tra cơ bản!")


# ============================================================
# 18. THỐNG KÊ MÔ TẢ DỮ LIỆU SAU KHI LÀM SẠCH
# ============================================================

print("\n========== THỐNG KÊ MÔ TẢ ==========")

display(df_clean.describe())


# ============================================================
# 19. LƯU DỮ LIỆU ĐÃ LÀM SẠCH THÀNH FILE CSV MỚI
# ============================================================

output_path = "/content/daily_web_traffic_clean.csv"

df_clean.to_csv(
    output_path,
    index=False,
    encoding="utf-8-sig"
)

print("\nĐã lưu dữ liệu sạch tại:")
print(output_path)


# ============================================================
# 20. HIỂN THỊ TOÀN BỘ DỮ LIỆU SAU KHI LÀM SẠCH
# ============================================================

display(df_clean)
```
